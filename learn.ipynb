# Indexing by rows & cols

```python
df["Animal"]        #returns all values in "Animal" col as a Series data structure
```

```python
set(df["Countries Found"])      #returns all unique value in "Countries Found"
```

```python
df[["Habitat", "Countries Found"]]        #returns all pairs of values in "Habitat" and "Country" col as another dataframe
```

# Filtering

```python
df[df["Predators"]== "Lions, Hyenas"]        
#find all animals with "Lions, Hyenas" as predators
```

```python
df[(df["Predators"]== "Lions, Hyenas") & (df["Diet"]!= "Herbivore")]        
#find all animals with "Lions, Hyenas" as predators and are not herbivores
```

```python
df[df["Countries Found"].str.startswith("A")]
#finds animals from countries starting with A
```

```python
df[df["Countries Found"].isin(["Africa", "Arctic"])]
#finds all animals from "Africa" and "Arctic"
```

```python
df.loc[df["Animal"]== "Naked Mole Rat"]
#pulls up location and data of "Naked Mole Rat"
```

# Editing data

```python
df.loc[df["Animal"]== "Aardvark", "Countries Found"] = "Asia"
df.loc[df["Animal"]== "Aardvark"]

#changes countries found of aardvarks to asia (doesn't modify actual csv)

```

```python
df=df.drop(0)       #drops 1st row (not in actual csv file)
df.head()
```
